```python
import pulp

def solve_min_tiles(n):
    # 예외 처리: 1x1 그리드는 빈칸 1개뿐이므로 타일이 0개 필요함
    if n == 1:
        return 0

    # 1. 빈칸(Uncovered) 위치 설정 (대각선 배치)
    uncovered_cells = set((i, i) for i in range(n))
    
    # 2. 그리드 내에서 생성 가능한 '모든 유효한 직사각형' 찾기
    rectangles = []
    for r1 in range(n):
        for r2 in range(r1, n):
            for c1 in range(n):
                for c2 in range(c1, n):
                    is_valid = True
                    for r in range(r1, r2 + 1):
                        for c in range(c1, c2 + 1):
                            if (r, c) in uncovered_cells:
                                is_valid = False
                                break
                        if not is_valid:
                            break
                    if is_valid:
                        rectangles.append((r1, r2, c1, c2))
                        
    # 3. 최적화 문제 정의 (Minimize)
    prob = pulp.LpProblem("Minimize_Tiles", pulp.LpMinimize)
    
    x = pulp.LpVariable.dicts("rect", rectangles, cat=pulp.LpBinary)
    prob += pulp.lpSum([x[rect] for rect in rectangles])
    
    for r in range(n):
        for c in range(n):
            if (r, c) not in uncovered_cells:
                covering_rects = [
                    rect for rect in rectangles 
                    if rect[0] <= r <= rect[1] and rect[2] <= c <= rect[3]
                ]
                prob += pulp.lpSum([x[rect] for rect in covering_rects]) == 1

    # 4. 풀이 실행
    prob.solve(pulp.PULP_CBC_CMD(msg=0))
    
    # 5. 안전장치: 최적해(Optimal)를 정상적으로 찾았는지 확인
    if prob.status == pulp.LpStatusOptimal:
        return int(pulp.value(prob.objective))
    else:
        # 솔버가 실패했을 경우 상태 코드 반환
        return f"계산 실패 (상태: {pulp.LpStatus[prob.status]})"

# === N값에 따른 검증 ===
print("수학적 공식 (2n - 2) 검증:")
for n in range(1, 2025):
    min_tiles = solve_min_tiles(n)
    print(f"Grid {n}x{n} -> 최소 타일 수: {min_tiles}개 (공식 결과: {2*n - 2}개)")
```
