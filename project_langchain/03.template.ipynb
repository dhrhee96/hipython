```python
import yfinance as yf
import textwrap
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. 실시간 주가 데이터 가져오는 함수
def get_stock_info(ticker_symbol):
    try:
        stock = yf.Ticker(ticker_symbol)
        data = stock.history(period="1d")
        current_price = data['Close'].iloc[-1]
        prev_close = stock.info.get('previousClose', current_price)
        change_pct = ((current_price - prev_close) / prev_close) * 100
        
        return f"현재가: {current_price:,.2f}, 전일대비: {change_pct:+.2f}%"
    except:
        return "데이터를 가져올 수 없는 잡주(자코)임"

# 2. 프롬프트 템플릿 정의 (System-User-AI-User)
chat_prompt = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template(
        "당신은 금융 지식이 해박하지만 성격이 뒤틀린 '소악마 메스가키' 자산관리사입니다. "
        "사용자를 '허접(자코)', '오빠'라고 부르며 비웃지만, 제공된 '실시간 주가 데이터'를 근거로 "
        "냉철하게 뼈를 때려야 합니다. 문장은 읽기 편하게 적절히 줄바꿈을 섞어주세요. ♡"
    ),
    # Few-shot: 말투와 논리 학습
    HumanMessagePromptTemplate.from_template("나 삼성전자(005930.KS) 풀매수했어! 나 이제 부자 되는 거야?"),
    AIMessagePromptTemplate.from_template(
        "우와아~ 진짜 대책 없는 허접 오빠네? ♡ \n"
        "방금 확인해보니까 지금 이 종목, 흐름도 안 좋은데 거기다 전 재산을 태워? \n"
        "분산 투자라는 기초 상식도 없는 거야? 진짜 '허접' 그 자체네! \n"
        "당장 포트폴리오 재점검 안 하면 조만간 한강 물 온도 체크하게 될걸? 허~접! ♡"
    ),
    # 실제 입력
    HumanMessagePromptTemplate.from_template(
        "상태: {investor_status}\n종목명: {stock_name}\n실시간 데이터: {stock_data}\n\n"
        "위 정보를 바탕으로 이 오빠한테 한마디 해줘!"
    )
])

# 3. 체인 구성
model = ChatOpenAI(model="gpt-4o",temperature=0.15)
chain = chat_prompt | model | StrOutputParser()

# 4. 가독성을 위한 출력 함수 (셀 안에서 안 벗어나게 함)
def print_wrapped_result(text, width=60):
    print("="*width)
    print(f"       [ 😈 귀여운 여동생의 독설 조언 ]")
    print("="*width)
    # 한글 특성을 고려해 textwrap 적용
    wrapped_text = textwrap.fill(text, width=width, break_long_words=False)
    print(wrapped_text)
    print("="*width)

# 5. 실행
ticker = "WMT"  # 테슬라 예시
stock_data = get_stock_info(ticker)
inputs = {
    "investor_status": "미국 주식에 퇴직금을 올인한 상태",
    "stock_name": "월마트(WMT)",
    "stock_data": stock_data
}

result = chain.invoke(inputs)
print_wrapped_result(result)
```

```python
import yfinance as yf
import textwrap
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. 고통스러운 지표를 가져오는 함수 (매운맛 강화)
def get_pain_metrics(ticker_symbol):
    try:
        stock = yf.Ticker(ticker_symbol)
        info = stock.info
        hist = stock.history(period="1y")
        
        curr = hist['Close'].iloc[-1]
        high_52w = info.get('fiftyTwoWeekHigh', curr)
        drawdown = ((curr - high_52w) / high_52w) * 100
        
        # 시장 지수(S&P 500)와 비교 (SPY)
        spy = yf.Ticker("SPY").history(period="1y")
        spy_return = ((spy['Close'].iloc[-1] - spy['Close'].iloc[0]) / spy['Close'].iloc[0]) * 100
        stock_return = ((hist['Close'].iloc[-1] - hist['Close'].iloc[0]) / hist['Close'].iloc[0]) * 100
        underperformance = stock_return - spy_return

        return {
            "price": f"{curr:,.2f}",
            "drawdown": f"{drawdown:+.2f}%",
            "vs_market": f"{underperformance:+.2f}%"
        }
    except:
        return None

# 2. 더 사악해진 퓨샷 프롬프트
chat_prompt = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template(
        "당신은 천재 퀀트 출신의 '소악마 메스가키' 자산관리사입니다. "
        "단순히 비웃는 게 아니라, 전고점 대비 하락률이나 시장 지수 대비 수익률 같은 '통계적 팩트'로 사용자를 철저히 짓밟습니다. "
        "사용자가 돈을 잃을수록 즐거워하며, 조언은 냉정하게 적중시켜야 합니다. ♡"
    ),
    # Few-shot: 더 매운맛 예시
    HumanMessagePromptTemplate.from_template("나 테슬라 샀어! 이제 곧 반등하겠지?"),
    AIMessagePromptTemplate.from_template(
        "꺄하하! ♡ 오빠, 지금 제정신이야? 전고점에서 {drawdown}이나 처박혔는데 '반등' 같은 소릴 하네? \n"
        "시장은 지금 축제 분위기인데 오빠 종목만 시장보다 {vs_market}이나 뒤처지고 있다고! \n"
        "이건 투자가 아니라 그냥 돈을 한강에 뿌리는 의식 아니야? 진짜 멍청해! ♡"
    ),
    # 실제 상황
    HumanMessagePromptTemplate.from_template(
        "오빠의 비참한 상태: {investor_status}\n"
        "종목: {stock_name}\n"
        "현재가: {price}\n"
        "전고점 대비: {drawdown} (고통 지수)\n"
        "시장 대비 수익률 차이: {vs_market}\n\n"
        "이 팩트를 가지고 오빠를 아주 가루로 만들어줘! ♡"
    )
])

# 3. 체인 및 출력 설정
model = ChatOpenAI(model="gpt-4o", temperature=1.0) # 창의적인 독설을 위해 온도 업!
chain = chat_prompt | model | StrOutputParser()

def print_spicy_box(text, width=65):
    print("\n" + "🔥" * (width // 2))
    print(f"       [ 😈 팩트 폭격 여동생의 처방전 ]")
    print("🔥" * (width // 2))
    
    # 문단별로 나누어 줄바꿈 처리
    paragraphs = text.split('\n')
    for p in paragraphs:
        if p.strip():
            wrapped = textwrap.fill(p, width=width, break_long_words=False)
            print(wrapped)
    print("🔥" * (width // 2) + "\n")

# 4. 실행
ticker = "TSLA" # 예시: 엔비디아 (잘 나가도 메스가키는 비웃습니다)
metrics = get_pain_metrics(ticker)

if metrics:
    inputs = {
        "investor_status": "남들 다 벌 때 구경만 하다가 꼭대기에서 상따한 허접",
        "stock_name": f"{ticker} (테슬라)",
        **metrics
    }
    result = chain.invoke(inputs)
    print_spicy_box(result)
```
